## Accessing data
### Accessing columns
To access columns use the `[colname]` or the `.`. This returns a Series.
For example, `adata['Height']` or `adata.Height`. In contrast, `adata[['Height']]` returns a DataFrame.


```python
#adata['Sepal.Length']
#adata.Sepal.Length # what is happening here? why? - various methods to access

adata[['Sepal.Length']] # put the first line of this cell into a different cell, and check the differences of the outputs.
```

### Accessing a subset of rows
To access a subset of rows, you can do the following: `adata[0:3]` selects rows 0, 1 and 2.

```python
adata[0:2]
```

### loc and iloc
A more straightforward and more powerful way to access parts of the DataFrame is with `loc` which takes a label or list of labels and `iloc` which takes an index or list of indices. Note: you're indexing not calling a function, so use `[]` not `()`.

```python
# adata.loc[[0, 2, 3], ['Sepal.Length', 'Species']]
# adata.loc[0:3, ['Sepal.Length', 'Species']]
adata.loc[3:5, ['Sepal.Length', 'Species']]
#adata.loc[:, ['Sepal.Length', 'Species']]
```

```python
# adata.iloc[:]
# adata.iloc[:, 1:4]
adata.iloc[3:6, [1,5]] # pay attention to the ending number -
```
